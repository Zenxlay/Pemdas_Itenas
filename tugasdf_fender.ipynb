```python
import pandas as pd

data = {
    'Nama': ['John', 'Jane', 'Bob', 'Alice'],
    'Usia': [25, 35, 30, 28],
    'Gaji': [50000, 60000, 70000, 55000]
}

df = pd.DataFrame(data)
```

```python
for i in range(len(df)):
    kenaikan_5 = lambda gaji: gaji * 1.05
    df.at[i, 'Gaji'] = kenaikan_5(df.at[i, 'Gaji'])
```

```python
print("\nData setelah kenaikan gaji 5% untuk semua karyawan: ")
print(df)
```

```python
for i in range(len(df)):
    if df.at[i, 'Usia'] > 30:
        kenaikan_2 = lambda gaji: gaji * 1.02
        df.at[i, 'Gaji'] = kenaikan_2(df.at[i, 'Gaji'])
```

```python
print("\nData setelah kenaikan tambahan 2% untuk usia diatas 30: ")
print(df)
```

```python
print("\nRingkasan Akhir:")
print("Semua karyawan mendapatkan kenaikan gaji 5%")
print("Karyawan berusia 30 tahun keatas mendapatkan tambahan kenaikan 2%")
```
